```python
# --- Extracting the input data from the txt file ---

with open("input_data_2024_day01.txt") as f:
	input_data = f.readlines()

print("Example input data raw:", input_data[:5]) 
print("Lines in input data:", len(input_data))
```

```python
# --- Part 1 ---

# Let's process the input data to convert it into a list of integers

left_nums = []
right_nums = []

for line in input_data:
	line = line.strip().strip("\n")  # Remove any leading/trailing whitespace characters
	if line:  # Check if the line is not empty
		line_nums = line.split("   ")
		left_nums.append(int(line_nums[0]))
		right_nums.append(int(line_nums[1]))

print("Left numbers:", left_nums[:5])
print("Right numbers:", right_nums[:5])
```

```python
# Let's sort the lists to have their respecive numbers in order from smaller to larger

left_nums.sort()
right_nums.sort()

print("Sorted left numbers:", left_nums[:5])
print("Sorted right numbers:", right_nums[:5])
```

```python
# Now let's find the absoulte difference between each pair of numbers and sum them up

differences_list = []

for left_n, right_n in zip(left_nums, right_nums):
	difference = abs(left_n - right_n)
	differences_list.append(difference)

total_difference = sum(differences_list)

print("Result 1 - Total absolute difference:", total_difference)
```

```python
# --- Part 2 ---

# Now we have to multiply how many times each number of the left list appears in the right list, and sum all those products

products_list = []

for left_n in left_nums:
	count_in_right = right_nums.count(left_n)
	product = left_n * count_in_right
	products_list.append(product)

total_product = sum(products_list)

print("Result 2 - Total sum of products:", total_product)
```
